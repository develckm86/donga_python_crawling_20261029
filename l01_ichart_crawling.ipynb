```python
import requests
url="https://www.ichart.kr/rank"
headers={"User-Agent":"Mozilla/5.0"}
res=requests.get(url, headers=headers)
res.encoding="utf-8"
#res.content #접속해서 받아온 html이 문자열로 출력됨
res.text
```

```python
%pip install beautifulsoup4
```

```python
from bs4 import BeautifulSoup
#res.text
# parser 파서 파싱 : 문자열인 데이터를 다른 형으로 변형 
soup=BeautifulSoup(res.text,"html.parser") # 문자열  => html 객체로 변환
# soup.select("css선택자") : 복수 []
# soup.select_one("css선택자") : 단수
lis=soup.select(".group.relative.w-full")
#복수 데이터를 사용하려면 반복 탐색 for in 
rows=[] 
for li in lis :
    row={
        "순위" : li.select_one(".btn-cta-medium").get_text(),
        "제목" : li.select_one(".body-contents-medium").get_text(),
        "가수" : li.select_one(".body-caption-small").get_text()
    }
    rows.append(row)
rows 
```

```python
%pip install pandas
```

```python
import pandas
df=pandas.DataFrame(rows)
df
```
