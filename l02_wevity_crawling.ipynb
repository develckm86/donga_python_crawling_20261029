```python
import requests
from bs4 import BeautifulSoup
import pandas

url="https://www.wevity.com/?c=find"
#요청해서 html 받기 
#html 파싱
#추출할 선택자 찾기
#추출해서 객체와 리스트로 만들기
#테이블로 생성!
res=requests.get(url,headers={"User-Agent":"Mozilla/5.0"})
res.encoding="utf-8"

bs=BeautifulSoup(res.text,"html.parser")
lis=bs.select(".main-section>.ms-list>.list>li:not(.top)")
rows=[]
for li in lis :
    row={
        "제목" : li.select_one(".tit>a").get_text(),
        "분야" : li.select_one(".tit>.sub-tit").get_text().replace("분야 : ",""),
        "조직" : li.select_one(".organ").get_text(),
        "day" : li.select_one(".day").contents[0].get_text(strip=True),
        "상태" : li.select_one(".day").contents[1].get_text(strip=True)
    }
    rows.append(row)
#rows
df=pandas.DataFrame(rows)
df
```
